```python
import pandas as pd
```

1- Read the CSV file into a pandas DataFrame and print it.

```python
df = pd.read_csv('products.csv')
df
```

2- Display the first 5 rows of the DataFrame.

```python
df.head()
```

3- Display the summary statistics of the DataFrame.

```python
df.describe()
```

4- Display the data types of each column of the DataFrame.

```python
df.dtypes
```

5- Filter: Display only the products where Price > 5000 or Rating > 4.5.

```python
df[(df['Price'] > 5000) | (df['Rating'] > 4.5)]
```

6- Sort: Sort the products by Price in descending order.

```python
df.sort_values(by='Price', ascending=False, inplace=True)
```

7- Handle Missing Values:

A- Check for missing values in the DataFrame.

B- Fill missing Price values with the average price.

C- Fill missing Stock values with the average stock.

```python
df.isnull().sum()
```

```python
df['Price'].fillna(df['Price'].mean(), inplace=True)
```

```python
df['Stock'].fillna(df['Stock'].mean(), inplace=True)
```

8- Add a new column:
Add a column called In_Stock that is True if Stock > 0, otherwise False.

```python
df['In_Stock'] = df['Stock'] > 0
df['In_Stock']
```

9- Add a new column:
Add a column called Total_Value that contains Price * Stock.

```python
df['Total_Value'] = df['Price'] * df['Stock']
```

10- Filter: Display products where Price > 3000 and Rating >= 4.0.

```python
df[(df['Price'] > 3000) & (df['Rating'] >= 4.0)]
```

11- Aggregation: Calculate the average, minimum, and maximum Price, and total Stock.

```python
print(df['Price'].mean())
print(df['Price'].min())
print(df['Price'].max())
print(df['Total_Value'].mean())
print(df['Total_Value'].min())
print(df['Total_Value'].max())
```

12- Add a new column:
Add a column called Price_Category:

Expensive if Price >= 10000

Medium if Price >= 3000

Cheap if Price < 3000

```python
def price_tier(price):
    if price >= 10000:
        return 'Expensive'
    elif price >= 3000:
        return 'Medium'
    else:
        return 'Cheap'

df['Price_Tier'] = df['Price'].apply(price_tier)
df['Price_Tier']
```

13- Rename a column:
Rename Product_Name to Name.

```python
print(df.columns)
df.rename(columns={'Product_Name': 'Name'}, inplace=True)
print(df.columns)
df.head()
```

14- Remove a column:
Remove the Product_ID column from the DataFrame.

```python
df.drop('Product_ID', axis=1, inplace=True)
df
```

15- Drop Duplicates:
Check for duplicate rows and remove them from the DataFrame.

```python
df.drop_duplicates(inplace=True)
df
```

16- Sort:
Sort the products by Rating in descending order.

```python
df.sort_values(by='Rating', ascending=False, inplace=True)
df
```

17- Save the modified DataFrame to a new CSV file called products_modified.csv.

```python
df.to_csv('products_modified.csv', index=False)
```
